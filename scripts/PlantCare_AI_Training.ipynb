```python
# Step 2: Download PlantVillage Dataset (Fast Direct Download)
import os
import urllib.request
import zipfile

DATA_DIR = './plantvillage'
ZIP_PATH = './plantvillage.zip'

if not os.path.exists(DATA_DIR):
    print('Downloading PlantVillage dataset (~800MB)...')
    # Standard public mirror
    url = 'https://data.mendeley.com/public-files/datasets/tywbtsjrjv/files/d5652a28-c1d8-4b76-97f3-72fb80f94efc/file_downloaded'
    urllib.request.urlretrieve(url, ZIP_PATH)
    print('Extracting dataset...')
    with zipfile.ZipFile(ZIP_PATH, 'r') as zip_ref:
        zip_ref.extractall(DATA_DIR)
    print('Dataset ready!')
else:
    print('Dataset already present at', DATA_DIR)
```

```python
# Step 3: Run the Complete Training Pipeline (15-20 Epochs with Early Stopping)
# If scripts/train_model.py is not in the Colab working directory, we download or write it:
import urllib.request
import os

os.makedirs('scripts', exist_ok=True)
if not os.path.exists('scripts/train_model.py'):
    # Clone or fetch from your repository:
    !curl -s -o scripts/train_model.py https://raw.githubusercontent.com/arshaanbaig69/se/main/scripts/train_model.py

!python scripts/train_model.py \
    --data_dir ./plantvillage \
    --output_dir ./weights \
    --arch convnext_tiny \
    --epochs 20 \
    --batch_size 32 \
    --lr 0.0003
```
